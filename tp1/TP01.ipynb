```python

import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

base_url = "https://catalog.data.gov/dataset"
datasets = []
page = 1

MAX_ROWS = 2000

while True:
    print(f"🔍 قراءة الصفحة {page} ...")
    url = f"{base_url}?page={page}"
    r = requests.get(url)
    if r.status_code != 200:
        print(" انتهى التحميل.")
        break

    soup = BeautifulSoup(r.text, "lxml")
    items = soup.select(".dataset-content")

    if not items:
        print(" تم الوصول لنهاية الصفحات.")
        break

    for item in items:
        if len(datasets) >= MAX_ROWS:
            print(f" تم الوصول للحد الأقصى: {MAX_ROWS} صف.")
            break

        title = item.find("h3", class_="dataset-heading").get_text(strip=True)
        desc = item.find("div", class_="notes").get_text(strip=True) if item.find("div", class_="notes") else "No description"
        link = "https://catalog.data.gov" + item.find("a")["href"]
        org = item.find("div", class_="organization-type").get_text(strip=True) if item.find("div", class_="organization-type") else "Unknown"

        datasets.append({
            "Title": title,
            "Description": desc,
            "Organization": org,
            "Link": link
        })

    if len(datasets) >= MAX_ROWS:
        break

    page += 1
    time.sleep(1)

df = pd.DataFrame(datasets)
df.to_csv("datasets.csv", index=False, encoding="utf-8-sig")

print(f"تم حفظ الملف  {len(df)}")
```
