```python
from bs4 import BeautifulSoup
import requests
import pandas as pd

base_url = "https://books.toscrape.com/catalogue/page-{}.html"

books_data = []

for page in range(1, 51):
    print(f" جاري استخراج الصفحة {page} ...")
    
    url = base_url.format(page)
    r = requests.get(url)
    soup = BeautifulSoup(r.text, "html.parser")
    
    books = soup.find_all("article", class_="product_pod")
    
    for book in books:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").get_text()
        availability = book.find("p", class_="instock availability").get_text(strip=True)
        
        rating = book.p["class"][1] if len(book.p["class"]) > 1 else "Not rated"
        
        img_url = book.find("img")["src"]
        img_url = "https://books.toscrape.com/" + img_url.replace("../", "")
        
        books_data.append({
            "Title": title,
            "Price": price,
            "Availability": availability,
            "Rating": rating,
            "Image_URL": img_url
        })

df = pd.DataFrame(books_data)

df.to_csv("books.csv", index=False, encoding="utf-8")

print("✅ تم استخراج البيانات كاملة!")
print("📌 عدد الكتب:", len(df))
print("📂 تم حفظ الملف باسم books.csv")
```
