```python
import pandas as pd


with open("restaurants.csv", "r", encoding="utf-8", errors="ignore") as f:
    lines = f.readlines()

if not lines[0].split(",")[0].isdigit():
    print("⚠️ Removing invalid first line:", lines[0])
    lines = lines[1:]

with open("restaurants_fixed.csv", "w", encoding="utf-8") as f:
    f.writelines(lines)

df_rest = pd.read_csv("restaurants_fixed.csv", header=None)

df_rest.columns = [
    "id",
    "name",
    "borough",
    "buildingnum",
    "street",
    "zipcode",
    "phone",
    "cuisinetype"
]

df_rest = df_rest.applymap(
    lambda x: x.replace("Â", "").replace("º","") if isinstance(x, str) else x
)

df_rest.to_csv("restaurants_clean.csv", index=False)

df_rest.head(), df_rest.shape
```

```python
import pandas as pd
from astrapy import DataAPIClient

API_ENDPOINT = "https://5e084319-b8b3-476a-8db7-db44f0a41031-us-east1.apps.astra.datastax.com"
TOKEN ="AstraCS:KYLTrfBhvpvhwHPZIQQnTCcZ:eb6fa626ca2f320f55b912589f4612ffdc798ce2c7ea952564a294da50a4526a"

client = DataAPIClient(TOKEN)
db = client.get_database(API_ENDPOINT)

t_grade = db.get_table("inspections_by_grade")

df = pd.read_csv("inspections_clean.csv")
records = df.to_dict(orient="records")

count = 0
for r in records:

    if pd.isna(r["Grade"]):
        continue

    t_grade.insert_one({
        "grade": r["Grade"],
        "idrestaurant": r["idRestaurant"],
        "inspectiondate": r["InspectionDate"],
        "violationcode": r["ViolationCode"],
        "violationdescription": r["ViolationDescription"],
        "score": int(r["Score"]) if not pd.isna(r["Score"]) else None,
        "criticalflag": r["CriticalFlag"]
    })

    count += 1
    if count % 1000 == 0:
        print(f"{count} rows inserted...")

print("✔ DONE inserting into inspections_by_grade")
```
